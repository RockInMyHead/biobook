## 1. Проверка LLM (LangChain + Ollama)

*Примечание: Перед запуском убедитесь, что модель скачана в контейнере Ollama, например: `docker exec -it <ollama-container-name> ollama run llama3`*

```python
from langchain_community.llms import Ollama

try:
    llm = Ollama(model="llama3", base_url="http://ollama:11434")
    response = llm.invoke("Explain what a DNA sequence is in one sentence.")
    print(response)
except Exception as e:
    print("Ошибка подключения: Убедитесь, что модель llama3 скачана в Ollama.", e)
```

## 3. Biopython - Анализ последовательностей

```python
from Bio.Seq import Seq

my_seq = Seq("AGTACACTGGT")
print("DNA Sequence:", my_seq)
print("Complement:", my_seq.complement())
print("Reverse Complement:", my_seq.reverse_complement())
print("Protein Translation:", my_seq.translate())
```

## 4. Scanpy - Single-Cell RNA-seq

```python
import scanpy as sc
import numpy as np
import anndata as ad

# Создаем фиктивные данные для демонстрации работы библиотеки
np.random.seed(42)
adata = ad.AnnData(np.random.poisson(1., size=(100, 2000)))
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=40)
sc.tl.umap(adata)
print("✅ Объект AnnData успешно создан и обработан ", adata)
```
